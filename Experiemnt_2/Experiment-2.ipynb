<b>Experiment 2.1</b>:  Write a Python program to perform basic text preprocessing on a given data. The program should:

Convert text to lowercase. (Also display how many uppercase?)

Remove punctuation.        (Also display how many different types of punctuations?)

Remove numbers.            (Also display how many numbers?)

Remove extra whitespace. (How much extra whitespace is present?)

<b>Experiment 2.2:</b> Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.

           (Three separate code: using NLTK, using spaCy  and without using NLTK/spacy)

<b>Experiment 2.3</b>: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
import re
import string 

def preprocess_text_file(input_file, output_file):
    try:
        with open(input_file, 'r', encoding = 'utf-8') as file:
            text = file.read()

        print(f"---Successfully read file: {input_file} ---")
        print(f"Original Text is: \n{text}")

        #1. convert to lowercase from uppercase
        uppercase_count = sum(1 for char in text if char.isupper())
        lowercase_text = text.lower()

        #2. Remove punctuation.(Also display how many different types of punctuations?)
        punctuation_in_text =  [char for char in lowercase_text if char in string.punctuation]
        unique_punctuations = set(punctuation_in_text)
        no_punc_text = lowercase_text.translate(str.maketrans('', '', string.punctuation))


        # 3. Count & Remove Numbers
        numbers_count = sum(1 for char in no_punc_text if char.isdigit())
        no_nums_text = re.sub(r'\d+', '', no_punc_text)
        
        # 4. Count & Remove Extra Whitespace
        # Match sequences of 2 or more whitespaces
        extra_spaces_matches = re.findall(r'\s{2,}', no_nums_text)
        extra_whitespace_count = sum(len(match) - 1 for match in extra_spaces_matches)
        
        # Accounting for any residual leading/trailing spaces outside blocks
        leading_trailing_spaces = max(0, (len(no_nums_text) - len(no_nums_text.strip())) - len(extra_spaces_matches))
        extra_whitespace_count += leading_trailing_spaces

        # Collapse all remaining whitespaces into single spaces
        cleaned_text = " ".join(no_nums_text.split())
        
        # Write the cleaned text to the output file
        with open(output_file, 'w', encoding='utf-8') as output_file:
            output_file.write(cleaned_text)
            
        # Display analysis metrics
        print("--- Preprocessing Metrics Report ---")
        print(f"• Uppercase letters found: {uppercase_count}")
        print(f"• Total punctuation marks found: {len(punctuation_in_text)}")
        print(f"• Different types of punctuations present: {len(unique_punctuations)} {list(unique_punctuations) if unique_punctuations else ''}")
        print(f"• Individual number digits found: {numbers_count}")
        print(f"• Extra whitespace characters removed: {extra_whitespace_count}")
        print(f"\n✅ Cleaned text successfully saved to: {output_file}")
        
    except FileNotFoundError:
        print(f"❌ Error: The file '{input_file}' was not found. Please check your path.")
    except Exception as e:
        print(f"❌ An error occurred: {e}")

# --- Execution ---
# Replace 'raw_data.txt' with your actual input file name or system path
input_path = "2.1_text_data.txt" 
output_path = "cleaned_data.txt"

preprocess_text_file(input_path, output_path)
```

# 1. using NLTK

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

# Download required tokenization models
nltk.download('punkt')
nltk.download('punkt_tab')

file_path = "cleaned_data.txt"

try:
    # Read text from file
    with open(file_path, 'r', encoding='utf-8') as file:
        text = file.read()
    
    # Tokenize
    sentences = sent_tokenize(text)
    words = word_tokenize(text)

    # Display Results
    print(f"--- NLTK Results from '{file_path}' ---")
    print(f"\n[SENTENCES - Total: {len(sentences)}]")
    for i, sent in enumerate(sentences, 1):
        print(f"  {i}: {sent}")

    print(f"\n[WORDS - Total: {len(words)}]")
    print(words)

except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found.")
```

# using spacy

```python
import spacy

# Load spaCy pipeline
nlp = spacy.load("en_core_web_sm")

file_path = "cleaned_data.txt"

try:
    # Read text from file
    with open(file_path, 'r', encoding='utf-8') as file:
        text = file.read()
    
    # Process text through pipeline
    doc = nlp(text)
    
    # Tokenize
    sentences = [sent.text for sent in doc.sents]
    words = [token.text for token in doc]

    # Display Results
    print(f"--- spaCy Results from '{file_path}' ---")
    print(f"\n[SENTENCES - Total: {len(sentences)}]")
    for i, sent in enumerate(sentences, 1):
        print(f"  {i}: {sent}")

    print(f"\n[WORDS - Total: {len(words)}]")
    print(words)

except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found.")
```

# without NLTK and spacy

```python
import re

file_path = "cleaned_data.txt"

try:
    # Read text from file
    with open(file_path, 'r', encoding='utf-8') as file:
        text = file.read()
    
    # 1. Sentence Tokenization
    sentence_raw = re.split(r'(?<=[.!?])\s+', text)
    sentences = [sent.strip() for sent in sentence_raw if sent.strip()]

    # 2. Word Tokenization
    words = re.findall(r"\w+(?:'\w+)?|[^\w\s]", text)

    # Display Results
    print(f"--- Pure Python Results from '{file_path}' ---")
    print(f"\n[SENTENCES - Total: {len(sentences)}]")
    for i, sent in enumerate(sentences, 1):
        print(f"  {i}: {sent}")

    print(f"\n[WORDS - Total: {len(words)}]")
    print(words)

except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found.")
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt')
nltk.download('stopwords')

# Read text from separate file
with open("cleaned_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Tokenize the text
tokens = word_tokenize(text)

# Get stop words
stop_words = set(stopwords.words("english"))

# Separate stop words and filtered tokens
stop_words_found = []
filtered_tokens = []

for word in tokens:
    if word.lower() in stop_words:
        stop_words_found.append(word)
    else:
        filtered_tokens.append(word)

# Display results
print("Original Text:")
print(text)

print("\nOriginal Tokens:")
print(tokens)

print("\nStop Words:")
print(stop_words_found)

print("\nFiltered Tokens:")
print(filtered_tokens)

# Store tokenized text in a separate file
with open("tokenized.txt", "w", encoding="utf-8") as file:
    file.write(" ".join(tokens))

# Store filtered text in another file
with open("filtered.txt", "w", encoding="utf-8") as file:
    file.write(" ".join(filtered_tokens))
```
