# Intelligent Consumer Complaint Classification & Prioritization
## Notebook 01: Dataset Acquisition, Validation & Initial Inspection

**Objective:**
Load and inspect the raw dataset downloaded from Hugging Face (`liri-uzh/cfpb-complaints-mini`) to verify data integrity, schema consistency, missingness, and target label distributions before building preprocessing and modeling pipelines.

```python
import pandas as pd
from pathlib import Path

# Define paths relative to the project root
data_dir = Path('../data/raw')
train_path = data_dir / 'train.csv'
test_path = data_dir / 'test.csv'

print(f"Train path exists: {train_path.exists()}")
print(f"Test path exists: {test_path.exists()}")
```

### 1. Load Datasets

```python
df_train = pd.read_csv(train_path)
df_test = pd.read_csv(test_path)

print(f"Train split dimensions: {df_train.shape}")
print(f"Test split dimensions:  {df_test.shape}")
df_train.head()
```

### 2. Schema and Missing Value Verification

```python
print("--- Train Info ---")
print(df_train.info())

print("\n--- Train Missing Values ---")
print(df_train.isnull().sum())

print("\n--- Test Missing Values ---")
print(df_test.isnull().sum())
```

### 3. Target Label Distribution

```python
train_dist = df_train['label'].value_counts(normalize=True).mul(100).round(2)
test_dist = df_test['label'].value_counts(normalize=True).mul(100).round(2)

dist_df = pd.DataFrame({
    'Train Count': df_train['label'].value_counts(),
    'Train %': train_dist,
    'Test Count': df_test['label'].value_counts(),
    'Test %': test_dist
})
dist_df
```

### 4. Text Length Distributions

```python
df_train['char_len'] = df_train['text'].astype(str).apply(len)
df_train['word_count'] = df_train['text'].astype(str).apply(lambda x: len(x.split()))

print("Train Character Length Summary:")
print(df_train['char_len'].describe())

print("\nTrain Word Count Summary:")
print(df_train['word_count'].describe())
```
